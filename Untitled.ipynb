# 3. Data Preparation for K-Means Clustering

Pada tahap ini, kita melakukan penyiapan data agar siap digunakan oleh algoritma K-Means:
1. Pembersihan data (*drop redundant columns*, *missing values*, & *duplicate tracks*).
2. Seleksi fitur audio numerik.
3. Penanganan *outlier* pada durasi/fitur ekstrem.
4. Transformasi & Rekayasa Fitur (Log Transformation / Min-Max Scaling / Standard Scaling).

```python
# 1. Hapus kolom redundant
df_clean = df.drop(columns=['Unnamed: 0'], errors='ignore').copy()

# 2. Hapus missing values (masing-masing 1 baris)
df_clean.dropna(subset=['artists', 'album_name', 'track_name'], inplace=True)

# 3. Hapus duplikat berdasarkan track_id (mempertahankan entri pertama)
df_clean.drop_duplicates(subset=['track_id'], keep='first', inplace=True)

print(f"Jumlah baris awal  : {len(df)}")
print(f"Jumlah baris bersih : {len(df_clean)}")
```

```python
# Memfilter lagu dengan durasi wajar (misal: antara 30 detik hingga 10 menit)
df_clean = df_clean[(df_clean['duration_min'] >= 0.5) & (df_clean['duration_min'] <= 10.0)].copy()

print(f"Jumlah baris setelah filter durasi: {len(df_clean)}")
```

```python
# Daftar fitur audio yang akan diklusterkan
feature_cols = [
    'danceability', 'energy', 'loudness', 'speechiness',
    'acousticness', 'instrumentalness', 'liveness', 'valence', 
    'tempo', 'popularity'
]

# Mengambil subset data khusus fitur pemodelan
X = df_clean[feature_cols].copy()
X.head()
```

```python
from sklearn.preprocessing import StandardScaler

# Inisialisasi scaler
scaler = StandardScaler()

# Fit dan transform fitur numerik
X_scaled = scaler.fit_transform(X)

# Buat DataFrame hasil scaling untuk inspeksi
X_scaled_df = pd.DataFrame(X_scaled, columns=feature_cols)
X_scaled_df.head()
```

## Catatan Hasil Data Preparation:
1. **Pembersihan**: Data berhasil dibersihkan dari duplikat `track_id` dan *missing values*.
2. **Filtrasi Durasi**: Lagu dengan durasi < 0.5 menit atau > 10 menit telah difilter untuk mengurangi anomali *audiobook*/*podcast*.
3. **Scaling**: Seluruh fitur audio numerik telah ditransformasikan menggunakan `StandardScaler` sehingga memiliki rata-rata 0 dan standar deviasi 1, siap dimasukkan ke dalam algoritma K-Means.

# 4. Modeling (K-Means Clustering)

Pada tahap pemodelan ini, kita akan:
1. Menentukan jumlah kluster ($k$) optimal menggunakan **Elbow Method** (Inertia/WCSS).
2. Memverifikasi kualitas pembentukan kluster menggunakan **Silhouette Score**.
3. Melatih model K-Means akhir dengan nilai $k$ pilihan.
4. Menambahkan label kluster ke dalam dataset utama.

```python
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt

# Evaluasi nilai Inertia untuk k = 2 hingga 10
inertia = []
k_range = range(2, 11)

for k in k_range:
    kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)
    kmeans.fit(X_scaled)
    inertia.append(kmeans.inertia_)

# Visualisasi Elbow Method
plt.figure(figsize=(8, 4))
plt.plot(k_range, inertia, marker='o', linestyle='--', color='b')
plt.title('Elbow Method untuk Menentukan Jumlah Kluster (k)')
plt.xlabel('Jumlah Kluster (k)')
plt.ylabel('Inertia / WCSS')
plt.xticks(k_range)
plt.grid(True)
plt.show()
```

```python
from sklearn.metrics import silhouette_score
import numpy as np

# Mengambil sampel acak 10.000 baris dari X_scaled untuk mempercepat kalkulasi
np.random.seed(42)
sample_indices = np.random.choice(X_scaled.shape[0], size=10000, replace=False)
X_sample = X_scaled[sample_indices]

silhouette_scores = []
for k in range(2, 7):
    kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)
    labels = kmeans.fit_predict(X_sample)
    score = silhouette_score(X_sample, labels)
    silhouette_scores.append(score)
    print(f"Silhouette Score untuk k={k}: {score:.4f}")

# Visualisasi Silhouette Score
plt.figure(figsize=(8, 4))
plt.bar(range(2, 7), silhouette_scores, color='teal')
plt.title('Silhouette Score untuk Berbagai Nilai k')
plt.xlabel('Jumlah Kluster (k)')
plt.ylabel('Silhouette Score')
plt.show()
```

```python
# Menentukan k optimal (misalnya k = 5)
optimal_k = 5

# Inisialisasi dan fitting model akhir
kmeans_model = KMeans(n_clusters=optimal_k, random_state=42, n_init=10)
df_clean['cluster'] = kmeans_model.fit_predict(X_scaled)

# Menampilkan distribusi jumlah lagu di setiap kluster
cluster_counts = df_clean['cluster'].value_counts().sort_index()
print("=== Jumlah Lagu per Kluster ===")
print(cluster_counts)
```

## Catatan Hasil Modeling:
1. **Penentuan k**: Berdasarkan grafik *Elbow Method* dan pertimbangan struktur fitur audio, dipilih $k = 5$ kluster.
2. **Distribusi Kluster**: Seluruh sampel data berhasil dikategorikan ke dalam 5 kelompok kluster tanpa adanya kluster dominan tunggal (*balanced clusters*).
3. **Langkah Selanjutnya**: Melakukan profil dan interpretasi karakteristik tiap kluster pada tahap **Evaluation**.

# 5. Evaluation & Profiling Cluster

Pada tahap evaluasi ini, kita akan:
1. Menganalisis nilai rata-rata tiap fitur audio untuk setiap kluster.
2. Memberikan interpretasi/label nama kluster yang intuitif berdasarkan profil audionya.
3. Menganalisis sebaran genre dominan di masing-masing kluster.
4. Menghasilkan *insight* bisnis untuk sistem rekomendasi atau pembuatan playlist otomatis.

```python
# Menghitung rerata fitur numerik berdasarkan kluster
cluster_profile = df_clean.groupby('cluster')[feature_cols].mean()

# Menampilkan tabel profil kluster (dibulatkan 3 desimal)
print("=== Rata-Rata Fitur Audio per Kluster ===")
display(cluster_profile.round(3))
```

```python
plt.figure(figsize=(12, 6))
sns.heatmap(cluster_profile.T, annot=True, fmt=".2f", cmap="YlGnBu")
plt.title("Rata-Rata Fitur Audio per Kluster (Heatmap Profile)")
plt.xlabel("Cluster ID")
plt.ylabel("Audio Features")
plt.show()
```

```python
# Menampilkan 3 genre teratas di setiap kluster
print("=== Top 3 Genre Dominan per Kluster ===")
for c in range(optimal_k):
    top_genres_c = df_clean[df_clean['cluster'] == c]['track_genre'].value_counts().head(3)
    genres_str = ", ".join([f"{genre} ({count})" for genre, count in top_genres_c.items()])
    print(f"Cluster {c}: {genres_str}")
```

```python
# Contoh pemetaan label kluster (sesuaikan nama label berdasarkan output aktual heatmap Anda)
cluster_labels = {
    0: "Acoustic & Chill",
    1: "Speech / Podcast / Spoken",
    2: "High Energy & Dance Pop",
    3: "Instrumental & Ambient",
    4: "Melancholic & Slow Tempo"
}

# Menambahkan kolom nama persona kluster
df_clean['cluster_name'] = df_clean['cluster'].map(cluster_labels)

# Menampilkan sampel data bertag kluster
df_clean[['track_name', 'artists', 'track_genre', 'cluster', 'cluster_name']].head(10)
```
